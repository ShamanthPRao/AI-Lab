```python
import random

# Display the board
def display_board(board):
    print("\n")
    print(" " + board[0] + " | " + board[1] + " | " + board[2])
    print("---+---+---")
    print(" " + board[3] + " | " + board[4] + " | " + board[5])
    print("---+---+---")
    print(" " + board[6] + " | " + board[7] + " | " + board[8])
    print("\n")


# Check whether a player has won
def check_winner(board, player):
    winning_combinations = [
        (0, 1, 2),  # Row 1
        (3, 4, 5),  # Row 2
        (6, 7, 8),  # Row 3
        (0, 3, 6),  # Column 1
        (1, 4, 7),  # Column 2
        (2, 5, 8),  # Column 3
        (0, 4, 8),  # Diagonal
        (2, 4, 6)   # Diagonal
    ]

    for combination in winning_combinations:
        if all(board[i] == player for i in combination):
            return True

    return False


# Check whether the board is full
def check_draw(board):
    return all(cell in ['X', 'O'] for cell in board)


# Human's move
def human_move(board):
    while True:
        try:
            position = int(input("Enter your position (1-9): "))

            if position < 1 or position > 9:
                print("Please enter a number between 1 and 9.")
            elif board[position - 1] in ['X', 'O']:
                print("That position is already occupied.")
            else:
                board[position - 1] = 'X'
                break

        except ValueError:
            print("Please enter a valid number.")


# Computer's move
def computer_move(board):
    available_positions = [
        i for i in range(9)
        if board[i] not in ['X', 'O']
    ]

    position = random.choice(available_positions)
    board[position] = 'O'

    print("Computer chose position:", position + 1)


# Main game
def play_game():
    board = ['1', '2', '3',
             '4', '5', '6',
             '7', '8', '9']

    print("===== TIC-TAC-TOE =====")
    print("You are X and the computer is O.")
    print("Choose a position from 1 to 9.")

    display_board(board)

    while True:
        # Human's turn
        human_move(board)
        display_board(board)

        if check_winner(board, 'X'):
            print("Congratulations! You win!")
            break

        if check_draw(board):
            print("The game is a draw!")
            break

        # Computer's turn
        computer_move(board)
        display_board(board)

        if check_winner(board, 'O'):
            print("Computer wins!")
            break

        if check_draw(board):
            print("The game is a draw!")
            break


# Start the game
play_game()
```

```python
# Vacuum Cleaner Agent with Tracing

def vacuum_cleaner(room_a, room_b, position):

    print("Initial State")
    print("----------------")
    print("Room A :", room_a)
    print("Room B :", room_b)
    print("Vacuum :", position)
    print()

    while room_a == "Dirty" or room_b == "Dirty":

        # Vacuum is in Room A
        if position == "A":

            if room_a == "Dirty":
                print("Action: SUCK Room A")
                room_a = "Clean"

            else:
                print("Action: MOVE from Room A to Room B")
                position = "B"

        # Vacuum is in Room B
        elif position == "B":

            if room_b == "Dirty":
                print("Action: SUCK Room B")
                room_b = "Clean"

            else:
                print("Action: MOVE from Room B to Room A")
                position = "A"

        print("Current State:")
        print("Room A :", room_a)
        print("Room B :", room_b)
        print("Vacuum :", position)
        print("----------------")

    print("\nGoal State Reached!")
    print("Both rooms are clean.")


# Initial state
room_a = "Dirty"
room_b = "Dirty"
position = "A"

# Run the agent
vacuum_cleaner(room_a, room_b, position)
```
