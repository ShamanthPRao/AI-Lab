```python
from collections import deque

# Initial state
initial = (5, 4, 6,
           1, 8, 7,
           3, 2, 0)

# Goal state
goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


def print_board(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


def get_moves(state):
    moves = []

    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    directions = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, name in directions:
        nr = row + dr
        nc = col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_blank = nr * 3 + nc

            new_state = list(state)
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            moves.append((tuple(new_state), name))

    return moves


def dfs(initial, goal):
    stack = [(initial, [])]
    visited = set()

    while stack:
        state, path = stack.pop()

        if state in visited:
            continue

        visited.add(state)

        if state == goal:
            return path

        for new_state, move in get_moves(state):
            if new_state not in visited:
                stack.append((new_state, path + [move]))

    return None


# Run DFS
solution = dfs(initial, goal)

if solution:
    print("Solution Found!")
    print("Number of moves:", len(solution))
    print("Moves:", solution)

    state = initial

    print("\nInitial State:")
    print_board(state)

    for i, move in enumerate(solution):
        for new_state, m in get_moves(state):
            if m == move:
                state = new_state
                break

        print("Step", i + 1, "-", move)
        print_board(state)

else:
    print("No solution found.")
```
